# Capstone Phase 1: Dataset Exploration & Statistical Analysis
**Project:** Intelligent Phishing Website Detection Using Machine Learning & Explainable AI  
**Dataset:** UCI Phishing Websites (11,055 instances, 30 features)

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load raw dataset
data_path = '../data/raw/phishing_dataset.csv'
df = pd.read_csv(data_path)
print('Dataset dimensions:', df.shape)
df.head()
```

```python
# Dataset Information & Null Verification
df.info()
print('Total missing values:', df.isnull().sum().sum())
```

```python
# Target Class Distribution
target_counts = df['Result'].value_counts()
print('Target Counts:\n', target_counts)
print('\nPercentage Breakdown:\n', df['Result'].value_counts(normalize=True) * 100)
```

```python
# Correlation with target
corr_target = df.corr()['Result'].drop('Result').sort_values(ascending=False)
print('Top 5 Phishing Indicators:\n', corr_target.head(5))
print('\nTop 5 Protective Features:\n', corr_target.tail(5))
```
