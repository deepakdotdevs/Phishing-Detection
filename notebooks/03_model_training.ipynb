# Capstone Phase 2: Supervised Model Implementation & Training
**Project:** Intelligent Phishing Website Detection Using Machine Learning & Explainable AI  
**Models:** Random Forest (Primary), Support Vector Machine (RBF), Artificial Neural Network (MLP)

```python
import os
import pandas as pd
import joblib
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neural_network import MLPClassifier

# Load training data
train_df = pd.read_csv('../data/processed/train.csv')
train_scaled_df = pd.read_csv('../data/processed/train_scaled.csv')

X_train = train_df.drop(columns=['target'])
y_train = train_df['target']
X_train_scaled = train_scaled_df.drop(columns=['target'])
print('Training records:', len(train_df))
```

```python
# Train Random Forest (Primary Model)
rf = RandomForestClassifier(n_estimators=100, max_depth=15, min_samples_split=4, random_state=42, n_jobs=-1)
rf.fit(X_train, y_train)
print('Random Forest trained successfully!')
```

```python
# Train SVM (RBF Kernel on Scaled Inputs)
svm = SVC(C=1.0, kernel='rbf', gamma='scale', probability=True, random_state=42)
svm.fit(X_train_scaled, y_train)
print('SVM trained successfully!')
```

```python
# Train Artificial Neural Network (MLP Classifier)
mlp = MLPClassifier(hidden_layer_sizes=(64, 32), activation='relu', max_iter=300, random_state=42)
mlp.fit(X_train_scaled, y_train)
print('MLP/ANN trained successfully!')
```
