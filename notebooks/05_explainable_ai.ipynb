**Project:** Intelligent Phishing Website Detection Using Machine Learning & Explainable AI  
**Components:** Global SHAP TreeExplainer, Beeswarm Summary, and Feature Attributions.

```python
import os
import pandas as pd
import numpy as np
import joblib
import shap
import matplotlib.pyplot as plt

# Load model and data
model = joblib.load('../models/random_forest.joblib')
test_df = pd.read_csv('../data/processed/test.csv')
X_test = test_df.drop(columns=['target'])
print('Model and test instances loaded successfully!')
```

```python
# Compute SHAP Values
explainer = shap.TreeExplainer(model)
sample = X_test.sample(200, random_state=42)
shap_values = explainer.shap_values(sample)
sv = shap_values[1] if isinstance(shap_values, list) else shap_values[:, :, 1]
print('SHAP values computed across test sample!')
```

```python
# SHAP Summary Plot
shap.summary_plot(sv, sample, max_display=10)
```
