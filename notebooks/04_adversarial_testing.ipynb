# Capstone Phase 3: Adversarial Evasion Attack Simulation
**Project:** Intelligent Phishing Website Detection Using Machine Learning & Explainable AI  
**Objective:** Formulate threat model, simulate feature manipulation, and quantify Evasion Success Rate (ESR).

```python
import os
import pandas as pd
import numpy as np
import joblib
from adversarial.evasion import PhishingEvasionAttacker, evaluate_evasion

# Load model and test dataset
model = joblib.load('../models/random_forest.joblib')
test_df = pd.read_csv('../data/processed/test.csv')
X_test = test_df.drop(columns=['target'])
y_test = test_df['target']
print(f'Test samples ready: {len(test_df)}')
```

```python
# Run Moderate Evasion Experiment
attacker = PhishingEvasionAttacker(attack_intensity='moderate')
res = evaluate_evasion(model, X_test, y_test, attacker)
print(f'Total Tested: {res["total_phishing_tested"]}')
print(f'Evaded Detection: {res["evaded_count"]}')
print(f'Evasion Success Rate: {res["evasion_success_rate"]*100:.2f}%')
print(f'Mean Confidence Drop: {res["avg_confidence_drop"]*100:.2f}%')
```
